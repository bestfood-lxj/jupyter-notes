```python
import matplotlib.pyplot as plt

figure, axes = plt.subplots()
Drawing_colored_circle = plt.Circle((8,8), 6,fill=False)

axes.set_aspect(1)

axes.add_artist(Drawing_colored_circle)
ax = plt.gca()
ax.set_xlim([1,15])
ax.set_ylim([1,15])
plt.plot([8],[8], marker='o', color='black')
plt.show()
```
